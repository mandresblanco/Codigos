```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
%matplotlib inline
```

```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_mod2.csv")
df.head()
```

```python
df.describe()
df.info()
```

```python
# Engine size as potential predictor variable of price
sns.regplot(x="CPU_frequency", y="Price", data=df)
plt.ylim(0,)

plt.show()
```

```python
# Engine size as potential predictor variable of price
sns.regplot(x="Screen_Size_inch", y="Price", data=df)
plt.ylim(0,)

plt.show()
```

```python
# Engine size as potential predictor variable of price
sns.regplot(x="Weight_pounds", y="Price", data=df)
plt.ylim(0,)

plt.show()
```

```python
from scipy import stats

pearson_coef, p_value = stats.pearsonr(df['CPU_frequency'], df['Price'])
print("The Pearson Correlation Coefficient is", pearson_coef, " with a P-value of P =", p_value)  


pearson_coef, p_value = stats.pearsonr(df['Screen_Size_inch'], df['Price'])
print("The Pearson Correlation Coefficient is", pearson_coef, " with a P-value of P =", p_value)  


pearson_coef, p_value = stats.pearsonr(df['Weight_pounds'], df['Price'])
print("The Pearson Correlation Coefficient is", pearson_coef, " with a P-value of P =", p_value)  
```

```python
for param in ['RAM_GB','CPU_frequency','Storage_GB_SSD','Screen_Size_inch','Weight_pounds','CPU_core','OS','GPU','Category']:
    pearson_coef, p_value = stats.pearsonr(df[param], df['Price'])
    print(param)
    print("The Pearson Correlation Coefficient for ",param," is", pearson_coef, " with a P-value of P =", p_value)
```

```python
# grouping results
df_gptest = df[['GPU','CPU_core','Price']]
grouped_test1 = df_gptest.groupby(['GPU','CPU_core'],as_index=False).mean()

grouped_pivot = grouped_test1.pivot(index='GPU',columns='CPU_core')
# grouped_pivot = grouped_pivot.fillna(0) #fill missing values with 0
print(grouped_pivot)

fig, ax = plt.subplots()
im = ax.pcolor(grouped_pivot, cmap='RdBu')

#label names
row_labels = grouped_pivot.columns.levels[1]
col_labels = grouped_pivot.index

#move ticks and labels to the center
ax.set_xticks(np.arange(grouped_pivot.shape[1]) + 0.5, minor=False)
ax.set_yticks(np.arange(grouped_pivot.shape[0]) + 0.5, minor=False)

#insert labels
ax.set_xticklabels(row_labels, minor=False)
ax.set_yticklabels(col_labels, minor=False)

#rotate label if too long
plt.xticks(rotation=90)

fig.colorbar(im)
plt.show()
```
